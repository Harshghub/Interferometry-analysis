```python
address = 'COM3'
timeout = 5000
import argparse
import sys
import time
import pyvisa
rm = pyvisa.ResourceManager('@py')
```

```python
port_num = address[3:]
resource_name = f"ASRL{port_num}::INSTR"
```

```python
device = rm.open_resource(resource_name, timeout=timeout)
```

```python
idn = device.query("*IDN?")
print(f"Device identified: {idn.strip()}")
```

```python
idn = device.write("CHN 1")
print(f"Device identified: {idn}")
```

```python
idn = device.write("FREQ 150e6")
print(f"Device identified: {idn}")
```

```python
idn = device.write("MODPMFREQ 10")
print(f"Device identified: {idn}")
```

```python
idn = device.write("LOCAL")
print(f"Device identified: {idn}")
```

```python
device.close()
```
